**Questão 4 - Transformações e ajustes**

1. Nesta questão armazenamos na variável map_scores um dicionário correspondendo o valor ao seu correspondente númerico e em seguida aplicamos a função ".map" na coluna "PerformanceScore" utilizando como parâmetro a variável definida acima.

2. No código abaixo a expressão "Salary * (1 + Absences / 100) " é utilizada tendo como paramêtro as colunas "Salary" e "Absences", retornando um valor que indica o custo total do funcionário (salário + ausências) para a empresa.

3. Métricas ajustadas e calculadas como traz a coluna "Cost_Index" ajudam os gestores a tomar melhores decisões; tornam os dados mais fidedignos no que tange a realidade da empresa e dos seus funcionários; permite observar o gasto real da empresa e adiciona um peso relacionado ao comportamento (ausências). 


```
# Questão 4.1
map_scores= {
    "Exceeds": 3,
    "Fully Meets": 2,
    "Needs Improvement": 1,
    "PIP": 0
}
df["PerformanceScore"]= df["PerformanceScore"].map(map_scores)
df[["Employee_Name", "PerformanceScore"]].head(10)

# Questão 4.2
df["Cost_Index"]= df["Salary"] * (1 + df["Absences"] / 100)
df[["Employee_Name", "Salary", "Absences", "Cost_Index"]].head(10)

# Questão 4.3 - Markdown


```
