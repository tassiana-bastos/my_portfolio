**Etapa 2 - Estruturação e persistência**

**2.1** - As conversões necessárias são:
- As colunas "Subscription Type", "Country", "Gender" e "Device" do tipo "objeto" para o tipo "categoria" - o objetivo desta conversão é obter análises mais rápidas e um menor consumo de memória. 
- As colunas "Join Date" e "Last Payment Date" do tipo "objeto" para o tipo "datetime" - o objetivo desta conversão é a possibilidade de calcular diferenças e obter uma análise mais rica. 

---------------------------------------------------------------------------------------------------------------------

Colunas úteis como filtros para futuros dashboards:
- Subscription Type: útil para analisar receitas, perfil etário e comportamento por tipo de assinatura;
- Country: útil para uma análise por região/mercado. Permite entender como a cultura/hábitos de um país interferem no consumo deste serviço;
- Gender: permite analisar se existe alguma preferência de plano por gênero;
- Device: permite uma análise de qual tipo de dispositivo é mais utilizado e pode ser utilizado para campanhas de marketing/produto;
- Age: facilita uma classificação por faixa etária, o que pode levar a uma análise dos perfis dos consumidores;

**2.3** Possível motivação dessa simulação:
Propor uma estratégia de precificação/valorização do produto baseado no perfil do consumidor. Clientes que utilizam mais estes equipamentos estão "on the go" e podem querer pagar um preço a mais pela versatilidade destes dispositivos e ainda podem querer pagar uma assinatura mais completa (Premium) por obter uma resolução melhor no Smartphone/Tablet por exemplo.


**2.4** Este tipo de estrutura permite análises mais pontuais e mais rápidas. Neste caso podemos verificar o comportamento dos clientes de um certo país e com uma certa assinatura de forma mais rápida e mais visual.

```
df["Subscription Type"]= df["Subscription Type"].astype("category")
df["Country"]= df["Country"].astype("category")
df["Gender"]= df["Gender"].astype("category")
df["Device"]= df["Device"].astype("category")

df["Join Date"] = pd.to_datetime(df["Join Date"], format="%d-%m-%y", errors="coerce")
df["Last Payment Date"] = pd.to_datetime(df["Last Payment Date"], format="%d-%m-%y", errors="coerce")

#---------------------------------------------------------------------------------------------------------
mais_recente= df["Last Payment Date"].max()
print("Última data de pagamento registrada", mais_recente)

mais_antiga = df["Last Payment Date"].min()
print("Primeira data de pagamento registrada:", mais_antiga)

ultimo_pagamento= pd.to_datetime("2023-07-15")
limite_6_meses= ultimo_pagamento - pd.DateOffset(months=6)

print("Considerando apenas pagamentos a partir de:", limite_6_meses.date())

usuarios_ativos= df[
    (df["Last Payment Date"] >= limite_6_meses) &
    (df["Last Payment Date"] <= mais_recente)
]

print("Dimensões do novo DataFrame:", usuarios_ativos.shape)

usuarios_ativos.to_csv("usuarios_ativos.csv", sep=";", encoding="utf-8", index=False)

display(usuarios_ativos.head())

#-------------------------------------------------------------------------------------------------------

df_modificado= df.copy()
dispositivos_moveis= ["Smartphone", "Tablet"]
df_modificado["Monthly Revenue"]= df_modificado["Monthly Revenue"].astype(float)
df_modificado.loc[df_modificado["Device"].isin(dispositivos_moveis), "Monthly Revenue"] *= 1.10
print("Após aumento de 10% para dispositivos móveis:")
print(df_modificado[["Device", "Monthly Revenue"]].head())
df_modificado.loc[df_modificado["Device"].isin(dispositivos_moveis), "Monthly Revenue"] /= 1.10

print("\nApós reverter a alteração:")
print(df_modificado[["Device", "Monthly Revenue"]].head())

#---------------------------------------------------------------------------------------------------------

df_multi= df.copy()
df_multi= df_multi.set_index(["Country", "Subscription Type"])
df_multi= df_multi.sort_index()
print("\nDataFrame com multi-index:")
print(df_multi.head())

exemplo= df_multi.loc[("Brazil", "Basic")]
print(exemplo.head())





```

**Etapa 4 -  Agrupamentos e agregações**

**4.1** Através destes agrupamentos podemos observar que em alguns países alguns tipos de assinaturas não são contemplados. Como exemplo temos a França cujo valor do plano Standard aparece como NaN e o Méximo cujo plano Premium também aparece como NaN, ou seja, não há um valor associado. Podemos observar também que no Dataset existe somente um tipo de duração de plano, que é de um mês, o que acaba limitando a análise.

**4.2** Após análise podemos observar que o grupo que consome mais é o que possue faixa etária "31-45", tanto em número de usuários quanto na soma da receita, ou seja, é o que traz mais lucro para a empresa. 



```
print(df["Plan Duration"].unique())
print(df["Plan Duration"].value_counts())

df["Plan Duration"]= df["Plan Duration"].astype(str)
df["Plan Duration"]= df["Plan Duration"].str.extract("(\d)").astype(int)

print(df["Plan Duration"].head())
print(df["Plan Duration"].dtypes)

agrupado= df.groupby(["Country", "Subscription Type"])["Plan Duration"].agg(["mean", "median"])

print(agrupado.head(20))
```

```
import numpy as np
import pandas as pd

@np.vectorize
def faixa_etaria(idade):
    if idade < 18:
        return"<18"
    elif 18 <= idade <= 30:
        return "18-30"
    elif 31 <= idade <= 45:
        return "31-45"
    elif 46 <= idade <= 60:
        return "46-60"
    else:
        return "60+"

df["Faixa Etária"]= faixa_etaria(df["Age"])
agrupado= df.groupby("Faixa Etária").agg(
    Media_Receita=("Monthly Revenue", "mean"), 
    Mediana_Receita=("Monthly Revenue", "median"), 
    Soma_Receita= ("Monthly Revenue", "sum"),
    Contagem_Usuarios=("User ID", "count")
)

print(agrupado)
```
